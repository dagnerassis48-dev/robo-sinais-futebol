```python

# 🤖 ROBÔ — SCANNER DE JOGOS DO DIA

from datetime import datetime

# Data de hoje
hoje = datetime.now().strftime("%Y-%m-%d")

url = "https://v3.football.api-sports.io/fixtures"

params = {
    "date": hoje
}

resposta = requests.get(
    url,
    headers=headers,
    params=params
)

jogos_dia = resposta.json()

print("📅 DATA:", hoje)
print("📡 STATUS:", resposta.status_code)
print("⚠️ ERROS:", jogos_dia.get("errors"))
print("⚽ TOTAL DE JOGOS:", jogos_dia.get("results"))

print("\n🏆 CAMPEONATOS ENCONTRADOS:\n")

campeonatos = {}

for jogo in jogos_dia.get("response", []):

    liga = jogo["league"]["name"]
    pais = jogo["league"]["country"]

    chave = f"{liga} - {pais}"

    if chave not in campeonatos:
        campeonatos[chave] = 0

    campeonatos[chave] += 1


for campeonato, quantidade in sorted(campeonatos.items()):
    print(f"{quantidade:>3} jogos | {campeonato}")
```

```python

# 🤖 ROBÔ — FILTRAR JOGOS QUE AINDA VÃO ACONTECER HOJE

from datetime import datetime, timezone

jogos_futuros = []

for jogo in jogos_dia.get("response", []):

    status = jogo["fixture"]["status"]["short"]

    # Jogos ainda não iniciados
    if status in ["NS", "TBD"]:

        casa = jogo["teams"]["home"]["name"]
        fora = jogo["teams"]["away"]["name"]
        liga = jogo["league"]["name"]
        pais = jogo["league"]["country"]

        horario = jogo["fixture"]["date"]

        jogos_futuros.append({
            "id": jogo["fixture"]["id"],
            "casa": casa,
            "fora": fora,
            "liga": liga,
            "pais": pais,
            "horario": horario
        })

print("📅 DATA:", hoje)
print("⚽ JOGOS NO DIA:", jogos_dia.get("results"))
print("🔜 JOGOS AINDA NÃO INICIADOS:", len(jogos_futuros))

print("\n🏆 PRÓXIMOS JOGOS:\n")

for jogo in jogos_futuros[:50]:

    horario = datetime.fromisoformat(
        jogo["horario"].replace("Z", "+00:00")
    )

    horario_brasil = horario.astimezone().strftime("%H:%M")

    print(
        f'{horario_brasil} | '
        f'{jogo["casa"]} x {jogo["fora"]} | '
        f'{jogo["liga"]} - {jogo["pais"]}'
    )
```

```python
# 🤖 ROBÔ — ORGANIZAR JOGOS DO DIA POR CAMPEONATO

from collections import Counter

contador_ligas = Counter()

for jogo in jogos_futuros:
    chave = f'{jogo["liga"]} - {jogo["pais"]}'
    contador_ligas[chave] += 1

print("🏆 CAMPEONATOS COM JOGOS HOJE:\n")

for campeonato, quantidade in contador_ligas.most_common():
    print(f"{quantidade:>3} jogos | {campeonato}")

print("\n📊 TOTAL DE CAMPEONATOS:", len(contador_ligas))
```

```python
# 🤖 ROBÔ — FILTRO DE COMPETIÇÕES

palavras_excluir = [
    "U17", "U18", "U19", "U20", "U21", "U22", "U23",
    "Youth", "Junior", "Reserve", "Reserves",
    "Women", "Woman", "Feminino",
    "Regional", "Amateur"
]

jogos_filtrados = []

for jogo in jogos_futuros:

    texto = (
        jogo["liga"] + " " +
        jogo["pais"]
    ).lower()

    excluir = False

    for palavra in palavras_excluir:
        if palavra.lower() in texto:
            excluir = True
            break

    if not excluir:
        jogos_filtrados.append(jogo)


print("⚽ JOGOS ANTES DO FILTRO:", len(jogos_futuros))
print("🟢 JOGOS PARA ANÁLISE:", len(jogos_filtrados))

print("\n🏆 EXEMPLOS DE CAMPEONATOS QUE SERÃO ANALISADOS:\n")

campeonatos_filtrados = Counter(
    f'{jogo["liga"]} - {jogo["pais"]}'
    for jogo in jogos_filtrados
)

for campeonato, quantidade in campeonatos_filtrados.most_common(40):
    print(f"{quantidade:>3} jogos | {campeonato}")
```

```python
# 🤖 ROBÔ — SELEÇÃO AUTOMÁTICA DE CANDIDATOS

from collections import defaultdict

# Agrupar jogos por campeonato
por_campeonato = defaultdict(list)

for jogo in jogos_filtrados:
    chave = f'{jogo["liga"]} - {jogo["pais"]}'
    por_campeonato[chave].append(jogo)

# Ordenar campeonatos pelo número de jogos
campeonatos_ordenados = sorted(
    por_campeonato.items(),
    key=lambda x: len(x[1]),
    reverse=True
)

candidatos = []

# No máximo 2 jogos por campeonato
for campeonato, jogos in campeonatos_ordenados:

    # Ordenar pelo horário
    jogos = sorted(jogos, key=lambda x: x["horario"])

    for jogo in jogos[:2]:
        candidatos.append(jogo)

# Limite para economizar as consultas da API
candidatos = candidatos[:30]

print("🎯 CANDIDATOS PARA ANÁLISE:", len(candidatos))

print("\n⚽ JOGOS SELECIONADOS:\n")

for i, jogo in enumerate(candidatos, 1):

    print(
        f'{i:02d}. {jogo["casa"]} x {jogo["fora"]} | '
        f'{jogo["liga"]} - {jogo["pais"]}'
    )
```

```python
# 🤖 ROBÔ — CORREÇÃO DO FILTRO DE TIMES

palavras_excluir_times = [
    "U17", "U18", "U19", "U20", "U21", "U22", "U23",
    "Youth", "Junior",
    "Women", "Woman", "Feminino",
    "Reserve", "Reserves"
]

jogos_filtrados = []

for jogo in jogos_futuros:

    texto = (
        jogo["liga"] + " " +
        jogo["pais"] + " " +
        jogo["casa"] + " " +
        jogo["fora"]
    ).lower()

    excluir = False

    for palavra in palavras_excluir_times:
        if palavra.lower() in texto:
            excluir = True
            break

    if not excluir:
        jogos_filtrados.append(jogo)


print("⚽ JOGOS ANTES DO FILTRO:", len(jogos_futuros))
print("🟢 JOGOS SÊNIORES PARA ANÁLISE:", len(jogos_filtrados))

print("\n🎯 PRIMEIROS CANDIDATOS:\n")

for i, jogo in enumerate(jogos_filtrados[:30], 1):
    print(
        f'{i:02d}. {jogo["casa"]} x {jogo["fora"]} | '
        f'{jogo["liga"]} - {jogo["pais"]}'
    )
```

```python

# 🤖 ROBÔ — FUNÇÃO DE ANÁLISE HISTÓRICA

def analisar_historico(team_id, casa_ou_fora):

    url = "https://v3.football.api-sports.io/fixtures"

    params = {
        "team": team_id,
        "season": 2026,
        "last": 10
    }

    resposta = requests.get(
        url,
        headers=headers,
        params=params
    )

    dados = resposta.json()

    jogos = []

    for jogo in dados.get("response", []):

        # Mandante
        if casa_ou_fora == "casa":
            if jogo["teams"]["home"]["id"] != team_id:
                continue

        # Visitante
        elif casa_ou_fora == "fora":
            if jogo["teams"]["away"]["id"] != team_id:
                continue

        # Somente jogos finalizados
        if jogo["fixture"]["status"]["short"] not in ["FT", "AET", "PEN"]:
            continue

        gols_casa = jogo["goals"]["home"]
        gols_fora = jogo["goals"]["away"]

        if gols_casa is None or gols_fora is None:
            continue

        if casa_ou_fora == "casa":
            gols_marcados = gols_casa
            gols_sofridos = gols_fora
        else:
            gols_marcados = gols_fora
            gols_sofridos = gols_casa

        total_gols = gols_casa + gols_fora

        jogos.append({
            "marcados": gols_marcados,
            "sofridos": gols_sofridos,
            "total": total_gols
        })

    if not jogos:
        return None

    return {
        "jogos": len(jogos),
        "over_1_5": sum(j["total"] >= 2 for j in jogos),
        "marcou": sum(j["marcados"] >= 1 for j in jogos),
        "sofreu": sum(j["sofridos"] >= 1 for j in jogos),
        "media_marcados": sum(j["marcados"] for j in jogos) / len(jogos),
        "media_sofridos": sum(j["sofridos"] for j in jogos) / len(jogos),
        "media_total": sum(j["total"] for j in jogos) / len(jogos)
    }

print("✅ Função de análise histórica criada!")
```

```python

# 🤖 FUNÇÃO AUTOMÁTICA — ANÁLISE CASA/FORA

def analisar_equipe_cordax(nome_equipe):

    url = "https://api.cordax.net/Fixtures"

    params = {
        "team": nome_equipe,
        "status": "FT",
        "from": "2026-01-01",
        "to": "2026-09-11"
    }

    resposta = requests.get(
        url,
        headers=headers_cordax,
        params=params
    )

    dados = resposta.json()

    if not isinstance(dados, list):
        return None

    # -------------------------
    # SEPARAR CASA E FORA
    # -------------------------

    jogos_casa = [
        jogo for jogo in dados
        if jogo.get("HomeTeam") == nome_equipe
    ]

    jogos_fora = [
        jogo for jogo in dados
        if jogo.get("AwayTeam") == nome_equipe
    ]

    # últimos 10
    jogos_casa = jogos_casa[-10:]
    jogos_fora = jogos_fora[-10:]

    def calcular(jogos, casa=True):

        if len(jogos) == 0:
            return None

        over_15 = 0
        marcou = 0
        sofreu = 0

        gols_marcados = 0
        gols_sofridos = 0
        total_gols = 0

        for jogo in jogos:

            gc = jogo.get("HomeScore")
            gf = jogo.get("AwayScore")

            if gc is None or gf is None:
                continue

            if casa:
                marcados = gc
                sofridos = gf
            else:
                marcados = gf
                sofridos = gc

            total = gc + gf

            if total >= 2:
                over_15 += 1

            if marcados >= 1:
                marcou += 1

            if sofridos >= 1:
                sofreu += 1

            gols_marcados += marcados
            gols_sofridos += sofridos
            total_gols += total

        n = len(jogos)

        return {
            "jogos": n,
            "over_15": over_15,
            "marcou": marcou,
            "sofreu": sofreu,
            "media_marcados": round(gols_marcados / n, 2),
            "media_sofridos": round(gols_sofridos / n, 2),
            "media_total": round(total_gols / n, 2)
        }

    return {
        "equipe": nome_equipe,
        "casa": calcular(jogos_casa, True),
        "fora": calcular(jogos_fora, False)
    }


# 🧪 TESTE
resultado = analisar_equipe_cordax("Drogheda United")

print("🤖 EQUIPE:", resultado["equipe"])

print("\n🏠 CASA")
print(resultado["casa"])

print("\n✈️ FORA")
print(resultado["fora"])
```

```python
from datetime import datetime

def ultimos_10_casa_fora(historico, nome_time):
    jogos_validos = []

    for jogo in historico:
        try:
            data = datetime.fromisoformat(
                jogo["FixtureDate"].replace("Z", "")
            )

            mandante = jogo["HomeTeam"] == nome_time
            visitante = jogo["AwayTeam"] == nome_time

            if not (mandante or visitante):
                continue

            gols_casa = int(jogo["HomeScore"])
            gols_fora = int(jogo["AwayScore"])

            jogos_validos.append({
                "data": data,
                "casa": jogo["HomeTeam"],
                "fora": jogo["AwayTeam"],
                "gols_casa": gols_casa,
                "gols_fora": gols_fora,
                "local": "CASA" if mandante else "FORA"
            })

        except:
            continue

    # Mais recente primeiro
    jogos_validos.sort(key=lambda x: x["data"], reverse=True)

    casa = [j for j in jogos_validos if j["local"] == "CASA"][:10]
    fora = [j for j in jogos_validos if j["local"] == "FORA"][:10]

    return casa, fora


casa, fora = ultimos_10_casa_fora(
    historico,
    nome_time
)

print("TIME:", nome_time)
print("\n🏠 ÚLTIMOS 10 EM CASA")

for j in casa:
    print(
        j["data"].date(),
        "|",
        j["casa"],
        "x",
        j["fora"],
        "|",
        j["gols_casa"],
        "-",
        j["gols_fora"]
    )

print("\n✈️ ÚLTIMOS 10 FORA")

for j in fora:
    print(
        j["data"].date(),
        "|",
        j["casa"],
        "x",
        j["fora"],
        "|",
        j["gols_casa"],
        "-",
        j["gols_fora"]
    )
```

```python
def calcular_estatisticas(lista_jogos, nome_time):

    total = len(lista_jogos)

    over15 = 0
    marcou = 0
    sofreu = 0

    gols_marcados = 0
    gols_sofridos = 0

    for j in lista_jogos:

        if j["local"] == "CASA":
            gm = j["gols_casa"]
            gs = j["gols_fora"]

        else:
            gm = j["gols_fora"]
            gs = j["gols_casa"]

        gols_marcados += gm
        gols_sofridos += gs

        if gm + gs >= 2:
            over15 += 1

        if gm >= 1:
            marcou += 1

        if gs >= 1:
            sofreu += 1

    return {
        "jogos": total,
        "over15": over15,
        "marcou": marcou,
        "sofreu": sofreu,
        "media_marcados": gols_marcados / total if total else 0,
        "media_sofridos": gols_sofridos / total if total else 0,
        "media_total": (gols_marcados + gols_sofridos) / total if total else 0
    }


estat_casa = calcular_estatisticas(casa, nome_time)
estat_fora = calcular_estatisticas(fora, nome_time)

print("📊", nome_time)

print("\n🏠 CASA")
print("Over 1.5:", estat_casa["over15"], "/ 10")
print("Marcou:", estat_casa["marcou"], "/ 10")
print("Sofreu:", estat_casa["sofreu"], "/ 10")
print("Média gols marcados:", round(estat_casa["media_marcados"], 2))
print("Média gols sofridos:", round(estat_casa["media_sofridos"], 2))
print("Média gols totais:", round(estat_casa["media_total"], 2))

print("\n✈️ FORA")
print("Over 1.5:", estat_fora["over15"], "/ 10")
print("Marcou:", estat_fora["marcou"], "/ 10")
print("Sofreu:", estat_fora["sofreu"], "/ 10")
print("Média gols marcados:", round(estat_fora["media_marcados"], 2))
print("Média gols sofridos:", round(estat_fora["media_sofridos"], 2))
print("Média gols totais:", round(estat_fora["media_total"], 2))
```

```python
import time

# Guarda o histórico de cada equipe para evitar consultas repetidas
cache_historico = {}

def buscar_historico_cordax(nome_time):

    # Se já buscamos esse time, reutiliza os dados
    if nome_time in cache_historico:
        return cache_historico[nome_time]

    url = "https://api.cordax.net/Fixtures"

    params = {
        "team": nome_time,
        "status": "FT"
    }

    resposta = requests.get(
        url,
        headers=headers_cordax,
        params=params
    )

    if resposta.status_code != 200:
        print("❌ Erro ao buscar:", nome_time)
        return []

    historico = resposta.json()

    cache_historico[nome_time] = historico

    print("✅", nome_time, "-", len(historico), "jogos")

    # Pequena pausa para não sobrecarregar a API
    time.sleep(1)

    return historico


print("🚀 Sistema automático preparado!")
print("Jogos de hoje:", len(jogos))
```

```python

# ==========================================
# ANÁLISE AUTOMÁTICA - OVER 1.5 GOLS
# ==========================================

resultados = []

for jogo in jogos:

    mandante = jogo["HomeTeam"]
    visitante = jogo["AwayTeam"]

    hist_mandante = historicos.get(mandante, [])
    hist_visitante = historicos.get(visitante, [])

    # Últimos 10 em casa e últimos 10 fora
    casa, _ = ultimos_10_casa_fora(hist_mandante, mandante)
    _, fora = ultimos_10_casa_fora(hist_visitante, visitante)

    # Estatísticas
    est_casa = calcular_estatisticas(casa, mandante)
    est_fora = calcular_estatisticas(fora, visitante)

    # Regra principal: mínimo 8/10 nos dois lados
    aprovada = (
        est_casa["over15"] >= 8 and
        est_fora["over15"] >= 8
    )

    # Classificação
    if est_casa["over15"] == 10 and est_fora["over15"] == 10:
        nivel = "🟢 A - MUITO FORTE"

    elif est_casa["over15"] == 10 and est_fora["over15"] >= 9:
        nivel = "🟢 B - FORTE"

    elif est_casa["over15"] >= 9 and est_fora["over15"] >= 8:
        nivel = "🟢 C - FORTE"

    elif aprovada:
        nivel = "🟡 APROVADA"

    else:
        nivel = "🔴 SEM ENTRADA"

    resultados.append({
        "jogo": f"{mandante} x {visitante}",
        "casa_over15": est_casa["over15"],
        "fora_over15": est_fora["over15"],
        "casa_marcou": est_casa["marcou"],
        "fora_marcou": est_fora["marcou"],
        "casa_sofreu": est_casa["sofreu"],
        "fora_sofreu": est_fora["sofreu"],
        "media_casa": round(est_casa["media_total"], 2),
        "media_fora": round(est_fora["media_total"], 2),
        "nivel": nivel
    })


# Mostrar resultado
print("=" * 70)
print("⚽ ANÁLISE AUTOMÁTICA — OVER 1.5 GOLS")
print("=" * 70)

for r in resultados:

    print("\n⚽", r["jogo"])
    print(
        f"🏠 Casa: Over 1.5 {r['casa_over15']}/10 | "
        f"Marcou {r['casa_marcou']}/10 | "
        f"Sofreu {r['casa_sofreu']}/10 | "
        f"Média total {r['media_casa']}"
    )

    print(
        f"✈️ Fora: Over 1.5 {r['fora_over15']}/10 | "
        f"Marcou {r['fora_marcou']}/10 | "
        f"Sofreu {r['fora_sofreu']}/10 | "
        f"Média total {r['media_fora']}"
    )

    print("➡️", r["nivel"])

print("\n" + "=" * 70)
print("✅ ANÁLISE DOS 23 JOGOS CONCLUÍDA")
print("=" * 70)
```

```python

# ==========================================
# FILTRO PROFISSIONAL — OVER 1.5
# ==========================================

ranking_final = []

for r in resultados:

    # Exige amostra completa
    if r["casa_over15"] < 8 or r["fora_over15"] < 8:
        continue

    # Índice de força
    indice = (
        r["casa_over15"] * 2 +
        r["fora_over15"] * 2 +
        r["casa_marcou"] +
        r["fora_marcou"] +
        r["casa_sofreu"] +
        r["fora_sofreu"] +
        (r["media_casa"] * 2) +
        (r["media_fora"] * 2)
    )

    # Classificação
    if (
        r["casa_over15"] >= 9 and
        r["fora_over15"] >= 9 and
        r["casa_marcou"] >= 8 and
        r["fora_marcou"] >= 7
    ):
        categoria = "🟢 MUITO FORTE"

    elif (
        r["casa_over15"] >= 8 and
        r["fora_over15"] >= 8 and
        r["casa_marcou"] >= 7 and
        r["fora_marcou"] >= 6
    ):
        categoria = "🟢 FORTE"

    else:
        categoria = "🟡 APROVADA"

    ranking_final.append({
        "jogo": r["jogo"],
        "indice": round(indice, 2),
        "categoria": categoria,
        "casa_over15": r["casa_over15"],
        "fora_over15": r["fora_over15"],
        "casa_marcou": r["casa_marcou"],
        "fora_marcou": r["fora_marcou"],
        "casa_sofreu": r["casa_sofreu"],
        "fora_sofreu": r["fora_sofreu"],
        "media_casa": r["media_casa"],
        "media_fora": r["media_fora"]
    })


# Ordenar pelo índice
ranking_final.sort(
    key=lambda x: x["indice"],
    reverse=True
)


print("=" * 75)
print("🏆 RANKING PROFISSIONAL — OVER 1.5 GOLS")
print("=" * 75)

for i, r in enumerate(ranking_final, 1):

    print(f"\n#{i} ⚽ {r['jogo']}")
    print(f"🔥 Índice: {r['indice']} | {r['categoria']}")

    print(
        f"🏠 Casa → Over {r['casa_over15']}/10 | "
        f"Marcou {r['casa_marcou']}/10 | "
        f"Sofreu {r['casa_sofreu']}/10 | "
        f"Média {r['media_casa']}"
    )

    print(
        f"✈️ Fora → Over {r['fora_over15']}/10 | "
        f"Marcou {r['fora_marcou']}/10 | "
        f"Sofreu {r['fora_sofreu']}/10 | "
        f"Média {r['media_fora']}"
    )


print("\n" + "=" * 75)
print("✅ FILTRO PROFISSIONAL CONCLUÍDO")
print("📌 Entradas:", len(ranking_final))
print("=" * 75)
```

```python

# ==========================================
# 📅 CORRIGIR DATAS DOS JOGOS
# API-FOOTBALL = DATA REAL
# ==========================================

import requests
import unicodedata
import re

def normalizar_nome(nome):
    nome = unicodedata.normalize(
        "NFKD", nome
    ).encode(
        "ASCII", "ignore"
    ).decode("ASCII")

    nome = nome.lower()
    nome = re.sub(r"[^a-z0-9]", "", nome)

    return nome


def buscar_jogos_api(data):

    url = "https://v3.football.api-sports.io/fixtures"

    params = {
        "date": data
    }

    resposta = requests.get(
        url,
        headers={
            "x-apisports-key": API_KEY
        },
        params=params
    )

    dados = resposta.json()

    if resposta.status_code != 200:
        print("❌ Erro API-Football:", resposta.status_code)
        return []

    return dados.get("response", [])


# Buscar os dois dias
jogos_api_12 = buscar_jogos_api("2026-09-12")
jogos_api_13 = buscar_jogos_api("2026-09-13")


print("=" * 65)
print("📅 CALENDÁRIO REAL")
print("=" * 65)

print("12/09 →", len(jogos_api_12), "jogos encontrados")
print("13/09 →", len(jogos_api_13), "jogos encontrados")


# Criar mapa dos jogos
mapa_datas = {}

for jogo in jogos_api_12:

    casa = normalizar_nome(
        jogo["teams"]["home"]["name"]
    )

    fora = normalizar_nome(
        jogo["teams"]["away"]["name"]
    )

    chave = casa + "_" + fora

    mapa_datas[chave] = "2026-09-12"


for jogo in jogos_api_13:

    casa = normalizar_nome(
        jogo["teams"]["home"]["name"]
    )

    fora = normalizar_nome(
        jogo["teams"]["away"]["name"]
    )

    chave = casa + "_" + fora

    mapa_datas[chave] = "2026-09-13"


print("\n" + "=" * 65)
print("🔎 DATAS DOS NOSSOS JOGOS")
print("=" * 65)


for jogo in jogos:

    casa = normalizar_nome(
        jogo["HomeTeam"]
    )

    fora = normalizar_nome(
        jogo["AwayTeam"]
    )

    chave = casa + "_" + fora

    data_real = mapa_datas.get(
        chave,
        "NÃO ENCONTRADO"
    )

    print(
        data_real,
        "→",
        jogo["HomeTeam"],
        "x",
        jogo["AwayTeam"]
    )
```

```python

# ==========================================
# 🔧 CORRIGIR NOMES DOS TIMES AUTOMATICAMENTE
# ==========================================

from difflib import SequenceMatcher

def similaridade(a, b):
    return SequenceMatcher(None, a, b).ratio()


def encontrar_data(casa_cordax, fora_cordax):

    melhor = None
    melhor_score = 0

    casa_c = normalizar_nome(casa_cordax)
    fora_c = normalizar_nome(fora_cordax)

    # Procurar nos jogos do dia 12
    for jogo_api in jogos_api_12:

        casa_a = normalizar_nome(jogo_api["teams"]["home"]["name"])
        fora_a = normalizar_nome(jogo_api["teams"]["away"]["name"])

        score_casa = similaridade(casa_c, casa_a)
        score_fora = similaridade(fora_c, fora_a)

        score_total = (score_casa + score_fora) / 2

        if score_total > melhor_score:
            melhor_score = score_total
            melhor = ("2026-09-12", jogo_api)

    # Procurar nos jogos do dia 13
    for jogo_api in jogos_api_13:

        casa_a = normalizar_nome(jogo_api["teams"]["home"]["name"])
        fora_a = normalizar_nome(jogo_api["teams"]["away"]["name"])

        score_casa = similaridade(casa_c, casa_a)
        score_fora = similaridade(fora_c, fora_a)

        score_total = (score_casa + score_fora) / 2

        if score_total > melhor_score:
            melhor_score = score_total
            melhor = ("2026-09-13", jogo_api)

    # Só aceita se a semelhança for boa
    if melhor and melhor_score >= 0.70:
        return melhor[0], melhor_score, melhor[1]

    return "NÃO ENCONTRADO", melhor_score, None


print("=" * 70)
print("📅 CALENDÁRIO CORRIGIDO")
print("=" * 70)

for jogo in jogos:

    data, score, jogo_api = encontrar_data(
        jogo["HomeTeam"],
        jogo["AwayTeam"]
    )

    if jogo_api:
        print(
            data,
            "→",
            jogo["HomeTeam"],
            "x",
            jogo["AwayTeam"],
            "| correspondência:",
            round(score * 100, 1),
            "%"
        )
    else:
        print(
            "❌ NÃO ENCONTRADO →",
            jogo["HomeTeam"],
            "x",
            jogo["AwayTeam"]
        )
```

```python
# 🔧 CORREÇÃO DEFINITIVA: nomes diferentes de times

def ultimos_10_casa_fora(historico, nome_time):
    jogos_validos = []

    nome_normalizado = normalizar_nome(nome_time)

    for jogo in historico:
        try:
            data = datetime.fromisoformat(
                jogo["FixtureDate"].replace("Z", "")
            )

            # Compara os nomes NORMALIZADOS
            mandante = (
                normalizar_nome(jogo["HomeTeam"]) == nome_normalizado
            )

            visitante = (
                normalizar_nome(jogo["AwayTeam"]) == nome_normalizado
            )

            if not (mandante or visitante):
                continue

            gols_casa = int(jogo["HomeScore"])
            gols_fora = int(jogo["AwayScore"])

            jogos_validos.append({
                "data": data,
                "casa": jogo["HomeTeam"],
                "fora": jogo["AwayTeam"],
                "gols_casa": gols_casa,
                "gols_fora": gols_fora,
                "local": "CASA" if mandante else "FORA"
            })

        except:
            continue

    jogos_validos.sort(
        key=lambda x: x["data"],
        reverse=True
    )

    casa = [
        j for j in jogos_validos
        if j["local"] == "CASA"
    ][:10]

    fora = [
        j for j in jogos_validos
        if j["local"] == "FORA"
    ][:10]

    return casa, fora


# 🧪 TESTE DO BODØ/GLIMT
hist_bodo = historicos["Bodø/Glimt"]

bodo_casa, bodo_fora = ultimos_10_casa_fora(
    hist_bodo,
    "Bodø/Glimt"
)

print("=" * 60)
print("🧪 TESTE FINAL — BODØ/GLIMT")
print("=" * 60)
print("Últimos 10 em CASA:", len(bodo_casa))
print("Últimos 10 FORA:", len(bodo_fora))

if len(bodo_casa) == 10:
    print("✅ CASA: OK")
else:
    print("⚠️ CASA:", len(bodo_casa))

if len(bodo_fora) == 10:
    print("✅ FORA: OK")
else:
    print("⚠️ FORA:", len(bodo_fora))
```

```python
# 📊 RECONSTRUIR ANÁLISE CASA/FORA

from datetime import datetime

def ultimos_10_casa_fora_novo(historico, nome_time):

    jogos = []

    for jogo in historico:

        try:
            if jogo["HomeTeam"] == nome_time:
                local = "CASA"
                gols_marcados = int(jogo["HomeScore"])
                gols_sofridos = int(jogo["AwayScore"])

            elif jogo["AwayTeam"] == nome_time:
                local = "FORA"
                gols_marcados = int(jogo["AwayScore"])
                gols_sofridos = int(jogo["HomeScore"])

            else:
                continue

            data = datetime.fromisoformat(
                jogo["FixtureDate"].replace("Z", "")
            )

            jogos.append({
                "data": data,
                "local": local,
                "gm": gols_marcados,
                "gs": gols_sofridos
            })

        except:
            continue

    jogos.sort(key=lambda x: x["data"], reverse=True)

    casa = [j for j in jogos if j["local"] == "CASA"][:10]
    fora = [j for j in jogos if j["local"] == "FORA"][:10]

    return casa, fora


def estatisticas_novas(jogos):

    total = len(jogos)

    if total == 0:
        return {
            "jogos": 0,
            "over15": 0,
            "marcou": 0,
            "sofreu": 0,
            "media_marcados": 0,
            "media_sofridos": 0,
            "media_total": 0
        }

    over15 = sum(1 for j in jogos if j["gm"] + j["gs"] >= 2)
    marcou = sum(1 for j in jogos if j["gm"] >= 1)
    sofreu = sum(1 for j in jogos if j["gs"] >= 1)

    gm = sum(j["gm"] for j in jogos)
    gs = sum(j["gs"] for j in jogos)

    return {
        "jogos": total,
        "over15": over15,
        "marcou": marcou,
        "sofreu": sofreu,
        "media_marcados": gm / total,
        "media_sofridos": gs / total,
        "media_total": (gm + gs) / total
    }


resultados_dia = []

for jogo in jogos_finais:

    casa = jogo["casa"]
    fora = jogo["fora"]

    hist_casa = historicos.get(casa, [])
    hist_fora = historicos.get(fora, [])

    casa_10, _ = ultimos_10_casa_fora_novo(
        hist_casa, casa
    )

    _, fora_10 = ultimos_10_casa_fora_novo(
        hist_fora, fora
    )

    est_casa = estatisticas_novas(casa_10)
    est_fora = estatisticas_novas(fora_10)

    resultados_dia.append({
        "data": jogo["data"],
        "casa": casa,
        "fora": fora,

        "jogos_casa": est_casa["jogos"],
        "jogos_fora": est_fora["jogos"],

        "over_casa": est_casa["over15"],
        "over_fora": est_fora["over15"],

        "marcou_casa": est_casa["marcou"],
        "marcou_fora": est_fora["marcou"],

        "sofreu_casa": est_casa["sofreu"],
        "sofreu_fora": est_fora["sofreu"],

        "media_casa": est_casa["media_total"],
        "media_fora": est_fora["media_total"]
    })


print("=" * 70)
print("📊 ANÁLISE RECONSTRUÍDA —", DATA_ESCOLHIDA)
print("=" * 70)

for r in resultados_dia:

    if r["jogos_casa"] < 10 or r["jogos_fora"] < 10:
        status = "⚠️ HISTÓRICO INSUFICIENTE"
    elif r["over_casa"] >= 8 and r["over_fora"] >= 8:
        status = "🟢 APROVADA"
    else:
        status = "🔴 SEM ENTRADA"

    print(
        f"\n⚽ {r['casa']} x {r['fora']}"
        f"\n🎯 OVER 1.5: {r['over_casa']}/10 casa | {r['over_fora']}/10 fora"
        f"\n⚽ MARCOU: {r['marcou_casa']}/10 | {r['marcou_fora']}/10"
        f"\n🛡️ SOFREU: {r['sofreu_casa']}/10 | {r['sofreu_fora']}/10"
        f"\n📈 MÉDIA TOTAL: {r['media_casa']:.2f} | {r['media_fora']:.2f}"
        f"\n➡️ {status}"
    )
```

```python

# ==========================================
# 🤖 FUNÇÃO PRINCIPAL DO ROBÔ
# ==========================================

def executar_robo_diario():

    print("🤖 INICIANDO ROBÔ...")
    print("=" * 45)

    # 1️⃣ Buscar jogos já carregados
    print("📅 Jogos disponíveis:", len(jogos_finais))

    # 2️⃣ Encontrar candidatos com 10 jogos em casa/fora
    candidatos = []

    for jogo in jogos_finais:

        casa = jogo["casa"]
        fora = jogo["fora"]

        hist_casa = encontrar_historico(casa)
        hist_fora = encontrar_historico(fora)

        casa_10, _ = ultimos_10_casa_fora_robusto(
            hist_casa, casa
        )

        _, fora_10 = ultimos_10_casa_fora_robusto(
            hist_fora, fora
        )

        if len(casa_10) >= 10 and len(fora_10) >= 10:
            candidatos.append(jogo)

    print("🔎 Candidatos encontrados:", len(candidatos))

    # 3️⃣ Analisar candidatos
    resultados_auto = []

    for jogo in candidatos:

        try:
            resultado = analisar_jogo_novo(jogo)
            resultados_auto.append(resultado)

        except Exception as e:
            print("⚠️ Erro:", jogo["casa"], "x", jogo["fora"])
            print(e)

    # 4️⃣ Regra principal:
    # Over 1.5 em pelo menos 8 dos últimos 10
    aprovados = [
        r for r in resultados_auto
        if r["over_casa"] >= 8
        and r["over_fora"] >= 8
    ]

    # 5️⃣ Ordenar pelos melhores scores
    aprovados = sorted(
        aprovados,
        key=lambda x: x["score"],
        reverse=True
    )

    # 6️⃣ Selecionar no máximo 3
    top_finais = aprovados[:3]

    print("✅ Aprovados:", len(aprovados))
    print("🎯 Selecionados:", len(top_finais))
    print("=" * 45)

    # 7️⃣ Mostrar os selecionados
    for i, r in enumerate(top_finais, 1):

        print(
            f"{i}. {r['casa']} x {r['fora']}"
        )

        print(
            f"   ⚽ Over 1.5: "
            f"{r['over_casa']}/10 + "
            f"{r['over_fora']}/10"
        )

        print(
            f"   🏆 Score: {r['score']:.1f}"
        )

        print()

    # Guardar resultado para outras células
    return top_finais


print("✅ Função criada com sucesso!")
```
